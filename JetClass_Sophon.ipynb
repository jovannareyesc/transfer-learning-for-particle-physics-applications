```python
import awkward as ak
import numpy as np
import uproot

# Load ROOT file
file = uproot.open("val_5M/HToBB_120.root")
tree = file["tree"]

# Extract relevant branches
px = tree["part_px"].array()
py = tree["part_py"].array()
pz = tree["part_pz"].array()
E  = tree["part_energy"].array()
eta = tree["part_deta"].array()
phi = tree["part_dphi"].array()
charge = tree["part_charge"].array()
is_electron = tree["part_isElectron"].array()
is_muon = tree["part_isMuon"].array()
is_photon = tree["part_isPhoton"].array()

# Compute transverse momentum
pt = np.sqrt(px**2 + py**2)

# Stack particle 3D spatial positions: points = [eta, phi, pt]
points = ak.concatenate([eta[:, :, np.newaxis],
                         phi[:, :, np.newaxis],
                         pt[:, :, np.newaxis]], axis=-1)

# Stack particle features: you can add more if needed
features = ak.concatenate([
    charge[:, :, np.newaxis],
    is_electron[:, :, np.newaxis],
    is_muon[:, :, np.newaxis],
    is_photon[:, :, np.newaxis]
], axis=-1)

# Lorentz vectors: [E, px, py, pz]
lorentz_vectors = ak.concatenate([
    E[:, :, np.newaxis],
    px[:, :, np.newaxis],
    py[:, :, np.newaxis],
    pz[:, :, np.newaxis]
], axis=-1)

# Mask: 1 for real particles, 0 for padding (based on zero energy or pt)
mask = pt > 0  # you can change the condition if needed


```
